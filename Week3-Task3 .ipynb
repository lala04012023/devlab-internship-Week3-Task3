```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
orders = pd.read_csv(r"C:\Users\Lala\Downloads\archive\olist_orders_dataset.csv")
order_items = pd.read_csv(r"C:\Users\Lala\Downloads\archive\olist_order_items_dataset.csv")
reviews = pd.read_csv(r"C:\Users\Lala\Downloads\archive\olist_order_reviews_dataset.csv")
customers = pd.read_csv(r"C:\Users\Lala\Downloads\archive\olist_customers_dataset.csv")
```

## Cədvəllərin merge olunması

```python
df = ( orders.merge(customers, on="customer_id").merge(order_items, on="order_id").merge(reviews, on="order_id", how="left"))
```

```python
df
```

## Yalnız çatdırlımış sifarişlər

```python
df[df["order_status"] == "delivered"]
```

## Tarix sütünlarının Convert edilməsi

```python
df["order_purchase_timestamp"] = pd.to_datetime(df["order_purchase_timestamp"])
```

## Year-Month extract edilməsi

```python
df["YearMonth"] = df["order_purchase_timestamp"].dt.to_period("M")
```

```python
df
```

## Cohort Month yaradilmasi

```python
df["CohortMonth"] = ( df.groupby("customer_unique_id")['YearMonth'].transform("min"))
```

```python
df["CohortMonth"]
```

## CohortIndex hesablanması

```python
df["CohortIndex"] = ((df["YearMonth"].dt.year - df["CohortMonth"].dt.year) * 12 + (df["YearMonth"].dt.month - df["CohortMonth"].dt.month))

```

```python
df["CohortIndex"]
```

## Cohort cədvəlinin yaradılması

```python
cohort_data = (df.groupby(["CohortMonth", "CohortIndex"])["customer_unique_id"] .nunique().reset_index().sort_values(by='customer_unique_id',ascending=False))
```

```python
cohort_pivot = cohort_data.pivot(
    index="CohortMonth",
    columns="CohortIndex",
    values="customer_unique_id")
```

```python
cohort_pivot
```

```python
cohort_pivot.head()
```

## Retention cədvəlinin hesablanması

```python
retention = cohort_pivot.div(cohort_pivot.iloc[:, 0], axis=0)

retention
```

```python
average_retention = (retention.iloc[:, 1:] .mean().mean()) * 100

round(average_retention, 2)
```

```python
retention.iloc[:, 1:].mean()
```

## Orta çatdırılma müddəti

```python
df["delivery_days"] = (df["order_delivered_customer_date"] -df["order_purchase_timestamp"]).dt.days
average_delivery_time = df["delivery_days"].mean().round(2)

average_delivery_time
```

```python
df["delivery_days"] = (df["order_delivered_customer_date"] -df["order_purchase_timestamp"]).dt.days

average_delivery_time = df["delivery_days"].mean().round(2)

average_delivery_time
```

```python
bins = [0, 3, 7, 14, 21, 100]

labels = [
    "0-3 gün",
    "4-7 gün",
    "8-14 gün",
    "15-21 gün",
    "21+ gün"]

df["delivery_bucket"] = pd.cut(
    df["delivery_days"],
    bins=bins,
    labels=labels)
```

```python
review_analysis = ( df.groupby("delivery_bucket")["review_score"].mean().reset_index())

review_analysis
```

```python
df['review_score'].mean()
```

```python
products = pd.read_csv(r"C:\Users\Lala\Downloads\archive\olist_products_dataset.csv")

products.head()
```

```python
df = df.merge(
    products[["product_id", "product_category_name"]],
    on="product_id",
    how="left")
```

```python
electronics = [
    "eletronicos",
    "informatica_acessorios",
    "telefonia",
    "pcs"]

fashion = [
    "fashion_calcados",
    "fashion_bolsas_e_acessorios",
    "fashion_roupa_feminina",
    "fashion_roupa_masculina",
    "fashion_esporte",
    "fashion_underwear_e_moda_praia",
    "fashion_roupa_infanto_juvenil"]
```

```python
df["CategoryGroup"] = "Other"

df.loc[
    df["product_category_name"].isin(electronics), "CategoryGroup"] = "Electronics"

df.loc[
    df["product_category_name"].isin(fashion),"CategoryGroup"] = "Fashion"
```

```python
category_df = df[df["CategoryGroup"] != "Other"].copy()
```

```python
category_cohort = (
    category_df.groupby(
        ["CategoryGroup", "CohortMonth", "CohortIndex"])["customer_unique_id"] .nunique().reset_index())
```

```python
category_pivot = category_cohort.pivot_table(
    index=["CategoryGroup", "CohortMonth"],
    columns="CohortIndex",
    values="customer_unique_id")
```

```python
category_retention = category_pivot.div(category_pivot.iloc[:, 0],axis=0)
```

```python
electronics_retention = category_retention.loc["Electronics"]
```

```python
fashion_retention = category_retention.loc["Fashion"]
```

```python
comparison = pd.DataFrame({
    "Electronics": electronics_retention.mean(),
    "Fashion": fashion_retention.mean()})

comparison
```

```python
plt.figure(figsize=(10,6))

comparison.plot(kind="bar")

plt.title("Electronics və Fashion kateqoriyalarının orta Retention müqayisəsi")
plt.xlabel("Cohort Index")
plt.ylabel("Retention Rate")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

### Electronics və Fashion kateqoriyaları üzrə retention müqayisəsi göstərir ki, müştəri loyallığı məhsul kateqoriyasına görə dəyişə bilər.
### Əgər Electronics kateqoriyasında retention daha yüksəkdirsə, bu müştərilər platformaya təkrar alış üçün daha çox qayıdırlar.
### Əgər Fashion retention göstəricisi yüksəkdirsə, moda məhsullarının daha tez-tez təkrar alındığını göstərir.
### Bu analiz marketinq büdcəsinin hansı məhsul kateqoriyalarına yönəldilməsini müəyyən etməyə kömək edir.

### Əgər cohort-lardan biri digərlərindən daha yüksək retention göstərirsə, həmin dövrdə həyata keçirilən marketinq kampaniyaları və ya satış strategiyaları daha uğurlu hesab oluna bilər.
